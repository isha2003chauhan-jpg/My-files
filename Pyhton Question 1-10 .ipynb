# Q.1 Reverse the string using for loop.

```python
# Logic
input_string = "Python"
reverse_str = ""
for char in input_string:
    reverse_str= char + reverse_str
print(reverse_str)
```

```python
def reverse_string(input_string: str) -> str:
    """function to reverse a given string using for loop.
    Args:
        input_str(str): The string to be reversed.
    Returns :
        str: The reversed version of input string.
    Raises:
        TypeError : IF input is not a string.
    """
    # validate the input type
    if not isinstance(input_string,str):
        raise TypeError("Input must be a string")
    reversed_str = ""
    for char in input_string:
        reversed_str = char + reversed_str
    return reversed_str
# ----main execution----
if __name__ == "__main__":
    try:
        # taking input from the user
        user_input = input("Enter your string to reverse: ")
        #calling the function
        result = reverse_string(user_input)
        # display the reversed string 
        print(f"Reversed string: {result}")
    except TypeError as e:
        print (e)
    except Exception as e:
        print(e)
```

```python
class string:
    def reverse_string(text : str) -> str:
        """function to reverse a given string using for loop.
        Args:
            input_str(str): The string to be reversed.
        Returns :
            str: The reversed version of input string.
        Raises:
            TypeError : IF input is not a string.
        """
        if not isinstance(text,str):
            raise TypeError("Input must be a string")
        result = ""
        for char in text:
            result = char + result
        return result
# ----main execution----
if __name__ == "__main__":
    try:
        # taking input from the user
        text = input("Enter your string to reverse: ")
        print("Reversed string:",string.reverse_string(text))
    except TypeError as e:
        print (e)
    except Exception as e:
        print(e)
```

# Q.2 Find the largest number in the list using for loop

```python
# logic
Number_list = [1,23,12,43,5,64,6,7,8,7,5,43]
largest = Number_list[0]
for num in Number_list:
    if num > largest:
        largest=num
print(largest)
```

```python
def largest(numbers_list:list[float]) -> float:
        """Finds the largest number in a list
        Args:
            (numbers: list[float]): A list containing numeric values(int or float).
        Returns:
            float: The largest number in the list 
        Raises:
            TypeError: if input is not a list or contains non-numeric values.
            ValueError: if list is empty
        """
        if not isinstance(numbers_list,list):
            raise TypeError("Input must be a list of numbers of float or int")
        if not numbers_list:
            raise ValueError("List can not be empty")
        for num in numbers_list:
            if not isinstance(num,(int,float)):
                raise TypeError("All elements must be numeric")

        largest = numbers_list [0]
        for num in numbers_list:
            if num>largest:
                largest = num
        return largest

if __name__ == "__main__":
    try:
        user_input = input("Enter numbers in the list saperated by space int or float only: ")
        numbers_list = [float(item) for item in user_input.split()]
        # calling the function
        result = largest(numbers_list)
        print("Largest number in the list is :" , result)
    except (TypeError,ValueError) as e:
        print (e) 
    except Exception as e:
        print(e)
    
```

# Q.3 Take a input from user and find whether number is prime or not

```python
number=357
for i in range (2,(int(number**.5)+1)):
    if number % i == 0:
        print("Not a prime number")
        break
else:
    print("Prime number")    # to find underoot number**.5
```

```python
def prime_number(number:int) -> bool:
    """Function to find whether a number is prime number or not.
        Args:
           number(int): The number to check.
        Returns:
            bool: True If the number is prime, False otherwise.
        Raises:
            TypeError: if input is not an int.
            ValueError: if number is less than 2.
        """
    if not isinstance(number,int):
        raise TypeError("Input must be an integer")
    if number<2:
        raise ValueError("prime numbers are greater than or equal to 2")
    for i in range (2,(int(number**.5)+1)):
        if number % i == 0:
            return False
            break
    else:
        return True  # to find underoot number**.5

if __name__ == "__main__":
    try:
        user_input = input("Enter the number to check : ")
        if not user_input.strip().isdigit():
            raise TypeError("Input must be a digit")
        number = int(user_input)
        result = prime_number(number)
        print(result)
        #calling function
        if prime_number(number):
            print(f'{number} is a prime number')
        else:
            print(f'{number} is not a prime number')
    except (TypeError,ValueError) as e:
        print (e) 
    except Exception as e:
        print('Something went wrong')
    
```

# Q.5 Calculate factorial of a number

```python
# Logic
num=7
fact = 1
for i in range(1,num+1):
    fact = fact*i
print(fact)
```

```python
def factorial(num:int) -> int:
    """
    Calculate the factorial of a non-negative integer.
    
    Parameters:
        num(int): The number for the factorial is required.
    Returns:
        int: Factorial of the given number
    Raises:
        ValueError: if num is negative.
        TypeError: if num is not a integer.
    """
    if isinstance(num,bool):
        raise TypeError("Boolean values are not allowed")
    # check type of num
    if not isinstance(num,int):
        raise TypeError("Input must be an integer")
    # negative check
    if num < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    fact = 1
    for i in range(1,num+1):
        fact=fact*i
    return fact
# main execution
if __name__ == "__main__":
    try:
        number=int(input("Enter your number to find factorial:"))
        print("Factorial :",factorial(number))
    except TypeError as e:
        print("Error:",e)
    except ValueError as e:
        print(e)
```

# Q.6. Palindrome Check (String Problem)

```python
A Palindrome is a word,number or sequence that reads the same forward and backward.
Example: mom
madam
121
112211
```

```python
# Logic
string = "mom"
reverse_string = string[::-1]
print(string==reverse_string)
print(reverse_string)
```

```python
def is_palindrome(text: str) -> bool:
    """Checks whether a given string is a palindrome.
    Returns:
        Returns True if string is a palindrome False otherwise.
    Raises:
        TypeError: If input is not a string.
        ValueError: If input string is empty.
    """
    # 1. input must be a string
    if not isinstance(text,str):
        raise TypeError("Input must be a string")
    # 2. string must not be empty
    if text.strip() == "":     # .strip will remove the extra space or any extra characters
        raise ValueError("Input string can not be empty")
    else:
        text = text.strip().lower()
    # Reverse the string
    reverse_string = ""
    for ch in text:
        reverse_string = ch+reverse_string
    return text == reverse_string

# main execution
if __name__ == "__main__":
    try:
        user_input = input("Enter your string to check :")
        if is_palindrome(user_input):
            print(f"{user_input} is a palidrome.")
        else:
            print(f"{user_input} is not a palindrome")
    except(TypeError,ValueError) as e:
        print("Error:", e)
```

```python
from typing import List
def second_largest(nums: List[int]) -> int:
    """
    Finds the second largest unique number in a list without using sorting.
    
    Parameters:
        nums (List[int] ): A list of integers.
    Returns:
        int: The second largest unique value.
        
    Raises:
        int:The second largest unique value.
        
    Raises:
        TypeError : If input is not a list or elements are not intergers.
        ValueError: If the list has fewer than 2 unique elements.
    """
    #validate that the input is a list
    if not isinstance(nums,list):
        raise TypeErro("Input must be a list of integers")
    # validate that all elements in the list are integers:
    if any(not isinstance(current_number,int) for current_number in nums):
        raise TypeError ("All elements of list should be integers")
    # check list has atleast two elements
    if len(nums)<2:
        raise ValueError("list should contain atleast two arguments.")
    if nums[0]>nums[1]:
        largest = nums[0]
        second = nums[1]
    else:
        largest = nums[1]
        second = nums[0]
    for current_number in nums[2:] :
        if current_number>largest:
            second=largest
            largest = current_number
        elif current_number != largest and current_number > second:
            second = current_number
    if second == largest:
        raise  ValueError("List must contains atleast two unique elements")
    return second
# main execution
if __name__ == "__main__":
    numbers_list = [1,2,4,3,5,7,6,9,12]
    try:
        result = second_largest(numbers_list)
    except (TypeError,ValueError) as error:
        print(f'Error: {error}')
    except Exception as e :
        print(e)
    else:
        print("second largest : ",result)
```

```python
# Logic
n = 234
total = 0
while n>0:
    print("Number",n)
    total = total + n % 10
    print("removed digit",n%10)
    n = n // 10
    print("Number after change",n)
print(total)
```

```python
def sum_of_digits(n: int) -> int:
    """
    Calculate the sum of digits of a non-negative integer.
    
    Parameters:
        n(int) : The number whose digit will be summed.
    Returns:
        int: Sum of the digits.
    Raises:
        TypeError: if input is not an integer.
        ValueError: if input is negative.
    """
    if not isinstance(n,int):
        raise TypeError("Input must be an integer")
    if n<0:
        raise ValueError("Input must be a positive number")
    total = 0
    while n>0:
        digit = (n%10)
        total = total + digit
        n = n // 10
    return total
# main Execution
if __name__ == "__main__":
    try:
        num = int(input("Enter  your number:"))
        result = sum_of_digits(num)
        print(f"Sum of digits is {result}")
    except (TypeError,ValueError)as error:
        print("Error",error)
```

# Q.10 Write a function that removes duplicates from a list while maintaining the original order.

```python
# Logic
List = [100,2,3,9,4,4,8,5,6,7]
result = []
for i in List:
    if i not in result:
        result.append(i)
result
```

```python
from typing import List,Any
def remove_duplicates(List: List[Any]) -> List[Any]:
    """
    Remove duplicates from a list while maintaining original order.
    Parameters:
        List (List[Any]) : A list containing any type of elements.
    Returns:
        List[Any] : A new list with duplicates removed.
    Raises:
        TypeError : if input is not a list.
        ValueError : if the list is empty
    """
    # validate input is a list
    if not isinstance(List,list):
        raise TypeError("Input must be a list.")
    # validate is not empty
    if len(List)==0:
        raise ValueError ("List must not be empty")
    # validate List has only one element
    if len(List)==1:
        print("List has only one element, No duplicates to remove.")
        return List
    result = []
    
    for i in List:
        if i not in result:
            result.append(i)
    return result

# main Execution'
if __name__ == "__main__":
    try:
        user_input = input("Enter your list: ")
        numbers_list = [int(item) for item in user_input.split()]
        result = remove_duplicates(numbers_list)
        print("After removing duplicates:",result)
    
    except (TypeError,ValueError) as e:
        print("Error:",e)
    except Exception as error:
        print("Something went wrong:" , error)
```
