# 1. string.capitalize()

Converts the first character to uppercase and the rest to lowercase.

```python
string1 = "isha"
print(string1.capitalize())     #first letter becomes capital

string2 = "iSHA in Ducat"
print(string2.capitalize())     #first letter capital, remaining letters lowercase

string3 = "1234 isha"
print(string3.capitalize())     #Number is unchanged, first alphabet stays same place
```

# 2. string.casefold()

Converts the string to lowercase more strongly than lower(), useful for case-insensitive comparison.

```python
string = "Python"
print(string.casefold())   #Converts uppercase Letters to Lowercase
string = "DUCAT"
print(string.casefold())   #convert mixed case to lowercase
string = "Isha in Ducat"
print(string.casefold())
string = "Mitgefühl"
print(string.casefold())

string1 = "ISHA"
string2 = "isha"
print(string1.casefold()==string2.casefold())    #checks both string without case difference
```

# 3. string.center(width[, fillchar])

Returns a centered string using spaces or the given fill character.

```python
string = "python"
print(string.center(16))     #center string in width 16 using space

string = "isha"
print(string.center(16,"-"))  #center string in width 16 using -

string = "isha"
print(string.center(12,"*")) #center string in width 12 using *
```

# 28. string.maketrans(x[, y[, z]])

Creates a translation table for translate(). It can replace characters, delete characters, or use a dictionary mapping.
replace 1 time 1 character/word.  X = character to replace ,  Y = new character( same length as x) , Z = character to delete.

```python
table = str.maketrans("is", "at")
text = "Farmer is working really hard"
result = text.translate(table)
print(result)                    # i is replaced with a and s is replaced with t

translation_table = str.maketrans("abc", "123", "x")
text = "abcdex"
result = text.translate(translation_table)
print(result)                   # a, b, c are replaced with 1, 2, 3 and x is deleted

translation_table = str.maketrans("", "", "aei")
text = "hello, amazing people!"
result = text.translate(translation_table)
print(result)                   # Delete characters a, e, and i

translation_table = str.maketrans({"a": "@", "b": "8", "c": "("})
text = "abcde"
result = text.translate(translation_table)
print(result)                     # a becomes @, b becomes 8, and c becomes (
```

# 39. string.split(separator=None, maxsplit=-1)

Splits the string into a list.

```python
string = "Python is very easy"
print(string.split())                          # Split by spaces

string = "my favourite place to visit is_Rishikesh"
print(string.split(sep="_", maxsplit=-1))      # Split by underscore, all parts

string = "apple,banana, mango, guava,cherry"
print(string.split(","))                     # Split by comma, all parts

string = "apple,banana,cherry"
print(string.split(",", 1))                   # Split by comma, only 1 split from left
```

# 40. string.splitlines(keepends=False)

Splits the string at line breaks and returns a list.
it will  split if a string has newline \n,\r ,\r\n on base of this..keepends is optional, it default value is FALSE.if it is false then \n is not in option.if it is TRUE then \n will be in option and returns list.


```python
string = "Python\nJava\nC++"
print(string.splitlines())               # Split lines and remove line break characters

string = "Isha\nKhushi\nkishi"
print(string.splitlines(keepends=True))       # Split lines and keep line break characters

string = "isha1\r\nisha2"
print(string.splitlines())                  # Split Windows-style line break

string = "my name is isha"
print(string.splitlines())               # No line break, returns one item
```

# 41. string.startswith(prefix[, start[, end]])

Checks whether the string starts with the given prefix.

```python
string = "I am learning Python Programming"
print(string.startswith("I"))                   # Is string starting with I?
print(string.startswith("Python"))              # Is string starting with Python?
print(string.startswith("P", 16))             # Is P starting from index 16?
print(string.startswith("learning", 6))          # Is learning starting from index 6?
print(string.startswith("am", 3, 5))            # Is am starting between index 3 and 5?

string = "main.py"
print(string.startswith(("main", "test")))      # Is string starting with main or test?
```

# 43. string.swapcase()

Converts lowercase letters to uppercase and uppercase letters to lowercase.

```python
string = "PyThOn"
print(string.swapcase())            # Lowercase becomes uppercase and uppercase becomes lowercase

string = "Isha_Ducat"
print(string.swapcase())         # Swap case around underscore too

string = "HELLO world"
print(string.swapcase())           # HELLO becomes hello and world becomes WORLD
  
string = "Panda123"
print(string.swapcase())           # Numbers are unchanged
```

# 45. string.translate(table)

Returns a translated string using a translation table created by maketrans(). it can remove or replace multiple characters at a time.


```python
table = str.maketrans("is", "at")
text = "This is a cat"
print(text.translate(table))               # Apply table: i -> a and s -> t

translation_table = str.maketrans("abc", "123", "x")
text = "abcdex"
print(text.translate(translation_table))      # Replace a, b, c and delete x

translation_table = str.maketrans("", "", "oaei")
text = "hello, amazing world!!!!"
print(text.translate(translation_table))         # Delete o, a, e, and i

translation_table = str.maketrans({"a": "@", "b": "8", "c": "&"})
text = "abcde"
print(text.translate(translation_table))       # Use dictionary translation table
```

# 47. string.zfill(width)
